# Dynamic Subgraphs — Quickstart

A governed dynamic-graph runtime + model-agnostic SDK. From one prompt, a stable
Supervisor **plans, validates, runs, and records** a transient LangGraph workflow.

This notebook installs it from PyPI, takes your OpenAI key, and runs a real graph
end-to-end — including exact token usage and cost.

> **Tip:** open this in [Google Colab](https://colab.research.google.com/) and run all cells — no local setup needed.

```python
# The [cost] extra adds LiteLLM so result.cost is computed automatically.
%pip install -q "dynamic-subgraphs[openai,cost]"
```

```python
# Paste your OpenAI key when prompted (it isn't stored to disk).
import getpass
import os

if not os.environ.get("OPENAI_API_KEY"):
    os.environ["OPENAI_API_KEY"] = getpass.getpass("OpenAI API key: ")
```

```python
from dynamic_subgraphs import DynamicSubgraphs, EngineConfig, Model

# With the [cost] extra installed, result.cost is computed automatically
# (no prices to specify). result.usage tokens are always exact.
engine = DynamicSubgraphs(EngineConfig(model=Model("openai", "gpt-5.4-nano")))

result = engine.run(
    "Compare solar and wind power for a small town and recommend one, briefly."
)

print("status:", result.status, "| ok:", result.ok)
print("plan:", result.plan.graph_id, "| nodes:", [n.id for n in result.plan.nodes])
print("tokens:", result.usage.total_tokens, "| cost USD:", result.cost)
print()
print(list(result.values.values())[-1])
```
